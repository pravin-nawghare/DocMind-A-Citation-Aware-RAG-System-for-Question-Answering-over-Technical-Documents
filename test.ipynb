```python
# filter metadata
def get_source_metadata(metadata):
    dl_meta = metadata.get("dl_meta", {})

    doc_items = dl_meta.get("doc_items", [])
    origin = dl_meta.get("origin", {})

    pages = set()

    for item in doc_items:
        for prov in item.get("prov", []):
            page_no = prov.get("page_no")
            if page_no is not None:
                pages.add(page_no)

    return {
        "source": origin.get(
            "filename",
            metadata.get("source")
        ),
        "pages": sorted(pages),
        "heading": (
            dl_meta.get("headings", [None])[0]
            if dl_meta.get("headings")
            else None
        ),
    }

# only keep this metadata in vector db
doc = ""
filtered_metadata = get_source_metadata(doc.metadata)

print(filtered_metadata)
```
